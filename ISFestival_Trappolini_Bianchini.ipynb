## Passo 2: Prepariamo gli strumenti 📦

Importiamo quello che ci serve per parlare con l'AI.

```python
from transformers import pipeline
import warnings
warnings.filterwarnings('ignore')

print("✅ Librerie caricate!")
```

## Passo 3: Scarichiamo il modello 🧠

Useremo **Llama** di Meta: un modello potente e completamente gratuito!

⏱️ Questo passaggio richiede 1-2 minuti (sta scaricando il modello).
È normale, rilassati! ☕

```python
print("🌐 Mi sto connettendo a Hugging Face...")
print("📥 Inizio download del modello Qwen 2.5...")
print("⏳ Questo richiederà 1-3 minuti. Sii paziente!")
print("=" * 60)

# Creiamo il nostro chatbot usando Qwen 2.5
chatbot = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    device_map="auto",
)

print("=" * 60)
print("✅ MODELLO CARICATO CON SUCCESSO!")
print("🎉 Il tuo chatbot è pronto!")
print()
print("📊 Informazioni sul modello:")
print(f"   - Nome: Qwen 2.5 Instruct (1.5B)")
print(f"   - Creatore: Alibaba Cloud")
print(f"   - Parametri: ~1.5 miliardi")
print(f"   - Tipo: Instruction-tuned (addestrato per conversazioni)")
print(f"   - Lingue: Multilingua (oltre 29 lingue!)")
```

## Passo 4: La nostra prima domanda! 💬

Proviamo a chiedere qualcosa all'AI!

```python
# Facciamo la prima domanda
domanda = "Ciao! Presentati in modo breve."

risposta = chatbot(
    domanda,
    max_new_tokens=100,
    do_sample=True,
    temperature=0.7
)

print(f"Tu: {domanda}")
print(f"AI: {risposta[0]['generated_text']}")
```

## 🤔 Cosa significano questi parametri?

- **max_new_tokens=100** → Quante parole può scrivere l'AI (circa)
- **temperature=0.7** → Quanto è "creativa" (0=noioso, 1=molto creativo)
- **do_sample=True** → Permette risposte più varie e naturali

Prova a cambiarli e vedi cosa succede! 🎨

## Passo 6: Creiamo una chat! 💭

Ora facciamo una vera conversazione interattiva!

```python
def chat_semplice():
    print("🤖 Chat attiva! (Scrivi 'esci' per uscire)\n")

    while True:
        domanda = input("Tu: ")

        if domanda.lower() == 'esci':
            print("👋 Ciao!")
            break

        risposta = chatbot(
            domanda,
            max_new_tokens=150,
            temperature=0.7
        )

        print(f"🤖 AI: {risposta[0]['generated_text']}\n")

# Avvia la chat!
chat_semplice()
```

## 🌐 BONUS: RAG con siti web!

Permetti all'AI di leggere contenuti da un sito web e usarli per rispondere!

## Chat che legge pagine web! 🚀

**Come funziona:**
1. Inserisci l'URL di un sito
2. L'AI scarica e legge il contenuto
3. Puoi fare domande sul contenuto del sito!

**Siti consigliati da provare:**
- Wikipedia (es: https://it.wikipedia.org/wiki/Intelligenza_artificiale)
- Articoli di blog
- Pagine informative

```python
!pip install requests beautifulsoup4 -q

import requests
from bs4 import BeautifulSoup

print("🌐 Chat con RAG da Web!")
url = input("🔗 URL del sito: ").strip()

# Scarica contenuto
try:
    soup = BeautifulSoup(requests.get(url, timeout=10).text, 'html.parser')
    for tag in soup(["script", "style"]): tag.decompose()
    contenuto = soup.get_text()[:3000]  # Primi 3000 caratteri
    print(f"✅ Caricato! ({len(contenuto)} caratteri)\n")
except:
    print("❌ Errore nel caricare il sito!")
    contenuto = ""

# Chat
while contenuto:
    q = input("Tu: ")
    if q.lower() == 'esci': break

    prompt = f"Contenuto web:\n{contenuto}\n\nDomanda: {q}\nRisposta:"
    r = chatbot(prompt, max_new_tokens=120, temperature=0.5, repetition_peanlty=1.1)
    print(f"🤖 AI: {r[0]['generated_text'].split('Risposta:')[-1].strip()}\n")
```
